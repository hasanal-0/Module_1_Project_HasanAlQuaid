## Handling duplicates: 

```python
def values_match(col1, col2):
    for a, b in zip(col1, col2):
        a_missing = pd.isna(a)
        b_missing = pd.isna(b)

        if a_missing and b_missing:
            continue
        elif a_missing != b_missing:
            return False
        elif a != b:
            return False

    return True


def remove_duplicate_columns(df):
    duplicate_mask = df.T.duplicated(keep="first")
    duplicate_pairs = []

    for i, is_duplicate in enumerate(duplicate_mask):
        if is_duplicate:
            for j in range(i):
                if values_match(df.iloc[:, i], df.iloc[:, j]):
                    duplicate_pairs.append(
                        (df.columns[i], df.columns[j])
                    )
                    break

    print("Duplicate column pairs:")
    for duplicate, original in duplicate_pairs:
        print(f"{duplicate}  ----  {original}")

    print("Number of columns dropped:", duplicate_mask.sum())

    return df.loc[:, ~duplicate_mask]


merge_3_df = remove_duplicate_columns(merge_3_df)
```

```python
# Upon looking at the data, these columns would be useful, so i am preping them to be used in a pairwise coorlatoin matrix in my EDA

merge_3_df["MOD_YEAR_x"] = merge_3_df["MOD_YEAR_x"].replace([9998, 9999], pd.NA)
merge_3_df["MOD_YEAR_x"] = pd.to_numeric(merge_3_df["MOD_YEAR_x"], errors="coerce")

merge_3_df["VSPD_LIM"] = merge_3_df["VSPD_LIM"].replace([98, 99], pd.NA)
merge_3_df["VSPD_LIM"] = pd.to_numeric(merge_3_df["VSPD_LIM"], errors="coerce")

merge_3_df["NUMOCCS"] = merge_3_df["NUMOCCS"].replace(99, pd.NA)
merge_3_df["NUMOCCS"] = pd.to_numeric(merge_3_df["NUMOCCS"], errors="coerce")

merge_3_df["AGE"] = merge_3_df["AGE"].replace([998, 999], pd.NA)
merge_3_df["AGE"] = pd.to_numeric(merge_3_df["AGE"], errors="coerce")

merge_3_df["TRAV_SP"] = merge_3_df["TRAV_SP"].replace([998, 999], pd.NA)
merge_3_df["TRAV_SP"] = pd.to_numeric(merge_3_df["TRAV_SP"], errors="coerce")
merge_3_df["TRAV_SP"] = merge_3_df["TRAV_SP"].replace(997, 152)

merge_3_df["HOUR_x"] = merge_3_df["HOUR_x"].replace(99, pd.NA) 
merge_3_df["HOUR_x"] = pd.to_numeric(merge_3_df["HOUR_x"], errors="coerce")
```
